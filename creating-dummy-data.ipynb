```python
import numpy as np

# Create a random number generator
rng = np.random.default_rng(0)

# Mean
mean = [-3, 0]

# Covariance matrix
cov = [
    [1.0, 0.8],
    [0.8, 1.0]
]

# Generate 500 random samples
data = rng.multivariate_normal(mean, cov, 500)

print("Shape of the data:", data.shape)
print(data)
```

```python
import matplotlib.pyplot as plt

plt.scatter(data[:, 0], data[:, 1])

plt.xlabel("x")
plt.ylabel("y")
plt.title("Scatter plot of dummy data")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Histogram of the first dimension
plt.hist(data[:, 0], bins=30)
plt.xlim(-7, 3)
plt.xlabel("x")
plt.ylabel("Frequency")
plt.title("Histogram of the first dimension")
plt.show()

# Histogram of the second dimension
plt.hist(data[:, 1], bins=30)
plt.xlim(-7, 3)
plt.xlabel("y")
plt.ylabel("Frequency")
plt.title("Histogram of the second dimension")
plt.show()
```

```python
# Problem 4: Addition of data

# Mean for the new data
mean_2 = [0, -3]

# Same covariance matrix
cov_2 = [
    [1.0, 0.8],
    [0.8, 1.0]
]

# Generate 500 new samples using the same rng
data_2 = rng.multivariate_normal(mean_2, cov_2, 500)

# Visualize both datasets
plt.scatter(data[:, 0], data[:, 1], label="0")
plt.scatter(data_2[:, 0], data_2[:, 1], label="1")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Scatter plot of two datasets")

plt.legend()

plt.show()
```

```python
# Problem 5: Data combination

combined_data = np.concatenate([data, data_2], axis=0)

print("Shape of combined data:", combined_data.shape)
```

```python
# Problem 6: Labeling

# Create labels
labels_0 = np.zeros(500)
labels_1 = np.ones(500)

# Combine the labels
labels = np.concatenate([labels_0, labels_1])

# Add labels as a new column
labeled_data = np.column_stack([combined_data, labels])

print("Shape of labeled data:", labeled_data.shape)
print(labeled_data)
```
