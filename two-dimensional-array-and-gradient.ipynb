```python
import numpy as np

# Create x values from -50 to 50 with intervals of 0.1
x = np.arange(-50, 50.1, 0.1)

# Calculate y
y = 0.5 * x + 1

print("x:", x)
print("y:", y)
print("Shape of x:", x.shape)
print("Shape of y:", y.shape)
```

```python
# Problem 2: Array combination

xy = np.column_stack((x, y))

print(xy)
print("Shape:", xy.shape)
```

```python
# Problem 3: Find the gradient

# Calculate the change in y
dy = np.diff(y)

# Calculate the change in x
dx = np.diff(x)

# Calculate the gradient
gradient = dy / dx

print("Gradient:", gradient)
print("Shape of gradient:", gradient.shape)
```

```python
# Problem 4: Draw graphs

import matplotlib.pyplot as plt

# We use the left endpoint of each interval: x[:-1]

# Graph 1: Linear function
plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Function: y = 0.5x + 1")
plt.grid()
plt.show()


# Graph 2: Approximate gradient
plt.figure(figsize=(8, 5))
plt.plot(x[:-1], gradient)
plt.xlabel("x")
plt.ylabel("Approximate gradient")
plt.title("Approximate Gradient")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


def compute_gradient(function, x_range=(-50, 50.1, 0.1)):
    """
    Calculate an approximate gradient from adjacent differences.
    """

    # Problem 1: create x and y
    array_x = np.arange(*x_range)
    array_y = function(array_x)

    # Problem 2: combine x and y
    array_xy = np.column_stack((array_x, array_y))

    # Problem 3: calculate approximate gradient
    dx = np.diff(array_x)
    dy = np.diff(array_y)
    gradient = dy / dx

    return array_xy, gradient
```

```python
import numpy as np
import matplotlib.pyplot as plt


def compute_gradient(function, x_range=(-50, 50.1, 0.1)):
    # Create x
    array_x = np.arange(*x_range)

    # Calculate y
    array_y = function(array_x)

    # Combine x and y
    array_xy = np.column_stack((array_x, array_y))

    # Calculate approximate gradient
    dx = np.diff(array_x)
    dy = np.diff(array_y)
    gradient = dy / dx

    return array_xy, gradient


# Function 1: y = x²
def function1(array_x):
    return array_x ** 2


# Calculate
array_xy1, gradient1 = compute_gradient(function1)

print("Array shape:", array_xy1.shape)
print("Gradient shape:", gradient1.shape)
print("Gradient:")
print(gradient1)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(array_xy1[:, 0], array_xy1[:, 1])

plt.xlabel("x")
plt.ylabel("y")
plt.title("y = x²")
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(array_xy1[:-1, 0], gradient1)

plt.xlabel("x")
plt.ylabel("Approximate gradient")
plt.title("Approximate Gradient of y = x²")
plt.grid()

plt.show()
```

```python
# Function 2: y = 2x² + 2x
def function2(array_x):
    return 2 * array_x ** 2 + 2 * array_x
```

```python
array_xy2, gradient2 = compute_gradient(function2)

print("Array shape:", array_xy2.shape)
print("Gradient shape:", gradient2.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(array_xy2[:, 0], array_xy2[:, 1])

plt.xlabel("x")
plt.ylabel("y")
plt.title("Function 2: y = 2x² + 2x")
plt.grid()
plt.show()
```

```python
# Function 3: y = sin(x/12)
def function3(array_x):
    return np.sin(array_x / 12)
```

```python
array_xy3, gradient3 = compute_gradient(function3)

print("Array shape:", array_xy3.shape)
print("Gradient shape:", gradient3.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(array_xy3[:, 0], array_xy3[:, 1])

plt.xlabel("x")
plt.ylabel("y")
plt.title("Function 3: y = sin(x/12)")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(array_xy1[:, 0], array_xy1[:, 1], label="y = x²")
plt.plot(array_xy2[:, 0], array_xy2[:, 1], label="y = 2x² + 2x")
plt.plot(array_xy3[:, 0], array_xy3[:, 1], label="y = sin(x/12)")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Functions 1, 2 and 3")
plt.legend()
plt.grid()
plt.show()
```
