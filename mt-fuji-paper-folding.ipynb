```python
# Problem 2: Function for arbitrary height

def calculate_folds(target_height, initial_thickness):
    folds = 0
    thickness = initial_thickness

    while thickness <= target_height:
        thickness *= 2
        folds += 1

    return folds


# Mt. Fuji
t0 = 0.00008
mt_fuji = 3776

print("Mt. Fuji:", calculate_folds(mt_fuji, t0), "folds")


# Proxima Centauri
proxima_centauri = 4.0175e16

print("Proxima Centauri:", calculate_folds(proxima_centauri, t0), "folds")
```

```python
# Problem 3: Required paper length

import math


def calculate_paper_length(target_height, initial_thickness):
    # Find the number of folds
    n = calculate_folds(target_height, initial_thickness)

    # Calculate the required paper length
    length = (math.pi * initial_thickness / 6) * \
             (2**n + 4) * (2**n - 1)

    return length


# Initial thickness
t0 = 0.00008

# Distances
moon = 384400000
mt_fuji = 3776
proxima_centauri = 4.0175e16


# Calculate required paper lengths
print("Mt. Fuji:", calculate_paper_length(mt_fuji, t0), "m")

print("Moon:", calculate_paper_length(moon, t0), "m")

print("Proxima Centauri:", calculate_paper_length(proxima_centauri, t0), "m")
```
