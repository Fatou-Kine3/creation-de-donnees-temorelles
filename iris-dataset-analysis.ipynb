```python
import pandas as pd
from sklearn.datasets import load_iris

print("Pandas:", pd.__version__)
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
```

```python
X = pd.DataFrame(
    iris.data,
    columns=["sepal_length", "sepal_width", "petal_length", "petal_width"]
)
```

```python
X.head()
```

```python
y = pd.Series(iris.target, name="Species")
```

```python
y.head()
```

```python
df = pd.concat([X, y], axis=1)
```

```python
df.head()
```

```python
df.head(4)
```

```python
# 1. First 4 samples
df.head(4)

# Check data types
df.info()

# 2. Number of samples for each label
df["Species"].value_counts()

# 3. Check missing values
df.isna().sum()

# 4. Mean, standard deviation, and quartiles
df.describe()
```

```python
# 1. Extract sepal_width in two ways
df["sepal_width"]

df.loc[:, "sepal_width"]
```

```python
# 2. Extract rows 50 through 99
df.loc[50:99]
```

```python
# 3. Extract petal_length for rows 50 through 99
df.loc[50:99, "petal_length"]
```

```python
# 4. Extract rows where petal_width is 0.2
df.loc[df["petal_width"] == 0.2]
```

```python
species_counts = df["Species"].value_counts()
```

```python
import matplotlib.pyplot as plt

plt.pie(
    species_counts,
    labels=species_counts.index,
    autopct="%1.1f%%"
)

plt.title("Number of samples per species")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
features = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

for feature in features:
    sns.boxplot(data=df, x="Species", y=feature)
    plt.title(f"Box plot of {feature}")
    plt.show()
```

```python
for feature in features:
    sns.violinplot(data=df, x="Species", y=feature)
    plt.title(f"Violin plot of {feature}")
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

features = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

for i in range(len(features)):
    for j in range(i + 1, len(features)):
        sns.scatterplot(
            data=df,
            x=features[i],
            y=features[j],
            hue="Species"
        )
        plt.title(f"{features[i]} vs {features[j]}")
        plt.show()
```

```python
sns.pairplot(
    df,
    hue="Species"
)

plt.show()
```

```python
correlation_matrix = df[features].corr()
```

```python
correlation_matrix
```

```python
sns.heatmap(
    correlation_matrix,
    annot=True
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix of Iris Features")
plt.show()
```
