```python
def calculate_survival_days(total_rice_grains, n_people):
    """Calculate how many days people can live with a given amount of rice.

    Parameters
    ----------
    total_rice_grains : int
        Total number of rice grains available.
    n_people : int
        Number of people sharing the rice.

    Returns
    -------
    int
        Number of days the rice can feed all people.
    """

    # According to FAO, we use 300 g of raw rice per adult per day.
    # USDA data gives about 20.5 g for 1,000 milled rice grains.
    # Therefore, one rice grain weighs about 0.0205 g.
    daily_rice_grams = 300
    grams_per_grain = 20.5 / 1000

    # Convert the daily rice requirement from grams to rice grains.
    daily_grains_per_person = daily_rice_grams / grams_per_grain

    # Calculate the total number of grains needed by all people per day.
    daily_grains_for_all = daily_grains_per_person * n_people

    # Calculate the number of complete days the rice can provide.
    survival_days = int(total_rice_grains / daily_grains_for_all)

    return survival_days
```

```python
def calculate_survival_days(total_rice_grains, n_people):
    """Calculate how many days a group can live on the available rice.

    Parameters
    ----------
    total_rice_grains : int
        Total number of rice grains available.
    n_people : int
        Number of people sharing the rice.

    Returns
    -------
    int
        Number of days the rice can feed all people.
    """

    # We assume that one person consumes 300 g of raw rice per day.
    # This value is based on FAO information.
    daily_rice_grams = 300

    # We assume that 1,000 rice grains weigh 20.5 g.
    # Therefore, one grain weighs 20.5 / 1000 g.
    grams_per_grain = 20.5 / 1000

    # Convert the daily rice requirement into number of grains.
    daily_grains_per_person = daily_rice_grams / grams_per_grain

    # Calculate the daily requirement for all people.
    daily_grains_for_all = daily_grains_per_person * n_people

    # Calculate the number of complete days the rice can feed them.
    survival_days = int(total_rice_grains / daily_grains_for_all)

    return survival_days
```

```python
# Calculate the rice received during the 75-day course.
list_n_grains, list_total_grains = compute_sorori_shinzaemon(n_days=75)

total_rice = list_total_grains[-1]

# Calculate how many days 4 people can live on this rice.
n_people = 4

survival_days = calculate_survival_days(total_rice, n_people)

print("Total rice grains:", total_rice)
print("Number of people:", n_people)
print("Number of days they can live:", survival_days)
```
