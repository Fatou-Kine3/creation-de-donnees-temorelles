```python
def create_wheat_board(n, m):
    n_squares = n * m
    
    wheat = [1]
    
    for _ in range(n_squares - 1):
        wheat.append(2 * wheat[-1])
    
    board = np.array(wheat, dtype=object).reshape(n, m)
    
    return board
```

```python
board_8x8 = create_wheat_board(8, 8)

print(board_8x8)
```

```python
total_wheat = np.sum(board_8x8)

print("Total number of wheat:", total_wheat)
```

```python
column_means = np.mean(board_8x8, axis=0)

print("Mean of each column:")
print(column_means)
```

```python
import matplotlib.pyplot as plt

plt.xlabel("column")
plt.ylabel("number")
plt.title("Average number of wheat in each column")

plt.bar(np.arange(1, 9), column_means)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create the 8x8 wheat board
def create_wheat_board(n, m):
    n_squares = n * m
    
    wheat = [1]
    
    for _ in range(n_squares - 1):
        wheat.append(2 * wheat[-1])
    
    return np.array(wheat, dtype=object).reshape(n, m)


# Create the 8x8 board
board_8x8 = create_wheat_board(8, 8)


# Create the heatmap
plt.xlabel("Column")
plt.ylabel("Row")
plt.title("Heatmap of wheat on the 8x8 board")

plt.pcolor(np.log2(board_8x8.astype(float)))

plt.colorbar(label="log2(number of wheat)")

plt.show()
```

```python
# First half: rows 0 to 3
first_half = board_8x8[0:4]

# Second half: rows 4 to 7
second_half = board_8x8[4:8]

# Calculate the total wheat in each half
first_half_total = np.sum(first_half)
second_half_total = np.sum(second_half)

# Calculate the ratio
ratio = second_half_total / first_half_total

print("First half:", first_half_total)
print("Second half:", second_half_total)
print("Ratio:", ratio)
```

```python
def create_wheat_board_append(n, m):
    n_squares = n * m
    
    board = np.array([1], dtype=np.uint64)
    
    for _ in range(n_squares - 1):
        board = np.append(board, np.uint64(2) * board[-1])
    
    return board.reshape(n, m)
```

```python
board_append = create_wheat_board_append(8, 8)

print("Board using np.append():")
print(board_append)
```

```python
def create_wheat_board_power(n, m):
    n_squares = n * m
    
    indices = np.arange(n_squares, dtype=np.uint64)
    
    board = np.uint64(2) ** indices
    
    return board.reshape(n, m)
```

```python
board_power = create_wheat_board_power(8, 8)

print("Board using exponentiation:")
print(board_power)
```

```python
print(np.array_equal(board_append, board_power))
```
