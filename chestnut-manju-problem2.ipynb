```python
import math
import matplotlib.pyplot as plt
```

```python
# Assumptions

MANJU_RADIUS = 0.02  # meters

# Volume of one chestnut manju, approximated as a sphere
MANJU_VOLUME = (4 / 3) * math.pi * MANJU_RADIUS ** 3

# Approximate radius of the Solar System: Neptune's orbit
AU = 149_597_870_700  # meters
SOLAR_SYSTEM_RADIUS = 30 * AU

# Approximate volume of the Solar System as a sphere
SOLAR_SYSTEM_VOLUME = (
    (4 / 3) * math.pi * SOLAR_SYSTEM_RADIUS ** 3
)

print("Manju volume:", MANJU_VOLUME, "m³")
print("Solar system volume:", SOLAR_SYSTEM_VOLUME, "m³")
```

```python
def calculate_doubling_time(initial_volume, target_volume, interval_minutes=5):
    """
    Calculate the time required for a growing volume
    to reach or exceed the target volume.

    Parameters
    ----------
    initial_volume : float
        Initial volume in m³.
    target_volume : float
        Target volume in m³.
    interval_minutes : int or float
        Doubling interval in minutes.

    Returns
    -------
    minutes : float
        Time required in minutes.
    doublings : int
        Number of doublings.
    """

    volume = initial_volume
    doublings = 0

    while volume < target_volume:
        volume *= 2
        doublings += 1

    minutes = doublings * interval_minutes

    return minutes, doublings
```

```python
minutes, doublings = calculate_doubling_time(
    MANJU_VOLUME,
    SOLAR_SYSTEM_VOLUME
)

days = minutes / (60 * 24)

print("Number of doublings:", doublings)
print("Time:", minutes, "minutes")
print("Time:", days, "days")
```

```python
volumes = []
volume = MANJU_VOLUME

for i in range(doublings + 1):
    volumes.append(volume)
    volume *= 2

times = [i * 5 for i in range(doublings + 1)]

plt.figure(figsize=(10, 6))

plt.plot(times, volumes)

plt.xlabel("Time (minutes)")
plt.ylabel("Total volume (m³)")
plt.title("Chestnut Manju Volume Growth")

plt.yscale("log")
plt.grid()

plt.show()
```

```python
# Example: a target volume of 1,000 m³

target_volume = 1000

minutes, doublings = calculate_doubling_time(
    MANJU_VOLUME,
    target_volume
)

print("Target volume:", target_volume, "m³")
print("Doublings:", doublings)
print("Time:", minutes, "minutes")
print("Time:", minutes / 60, "hours")
```

```python
## Assumptions and Sources

### Chestnut manju

A chestnut manju is approximated as an ellipsoid.

A Japanese chestnut manju product is listed as approximately 4 cm in
diameter and 3 cm in height.

Source:
Cake.jp, "栗饅頭 白あん":
https://cake.jp/item/3395601/

Therefore:

- Diameter = 4 cm = 0.04 m
- Height = 3 cm = 0.03 m

The volume is approximated using:

V = π/6 × diameter × diameter × height

### Solar System

For this simplified model, the Solar System is approximated as a sphere
extending from the Sun to Neptune.

NASA gives Neptune's average distance from the Sun as approximately
30.05 AU.

Source:
NASA, Solar System chapter:
https://science.nasa.gov/learn/basics-of-space-flight/chapter1-2/

One astronomical unit (AU) is approximately 150 million km.

Therefore:

- Solar System radius = 30.05 AU
- 1 AU ≈ 149,600,000 km
- 1 AU = 149,600,000,000 m

The volume of the Solar System is approximated using:

V = 4/3 × π × r³

### Simplifications

- The gaps between chestnut buns are ignored.
- Gravity is ignored.
- The buns are assumed to double in volume every 5 minutes.
- The Solar System is represented by a sphere, although its actual structure
  is much more complex.
```

```python
import math

# Chestnut manju dimensions
manju_diameter = 0.04  # m
manju_height = 0.03    # m

# Approximate manju as an ellipsoid
manju_volume = (
    math.pi / 6
    * manju_diameter
    * manju_diameter
    * manju_height
)

# Astronomical unit
AU = 149_600_000_000  # meters

# Approximate Solar System radius: Neptune's orbit
solar_system_radius = 30.05 * AU

# Approximate Solar System as a sphere
solar_system_volume = (
    4 / 3
    * math.pi
    * solar_system_radius ** 3
)

print("Manju volume:", manju_volume, "m³")
print("Solar System volume:", solar_system_volume, "m³")
```
