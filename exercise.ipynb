```python
import numpy as np
import pandas as pd

np.random.seed(123)
df1 = pd.DataFrame(
    {
        "x": np.random.uniform(size=50),
        "y": np.random.uniform(size=50),
        "target": np.repeat(0, 50),
    }
)
df2 = pd.DataFrame(
    {
        "x": np.random.uniform(-1, 0, size=50),
        "y": np.random.uniform(-1, 0, size=50),
        "target": np.repeat(1, 50),
    }
)
df = pd.concat([df1, df2], ignore_index=True)
df
```

```python
data0 = df.query("target == 0").loc[:, ["x", "y"]].values
target0 = df.query("target == 0").loc[:, "target"].values
data1 = df.query("target == 1").loc[:, ["x", "y"]].values
target1 = df.query("target == 1").loc[:, "target"].values
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.scatter(data0[:, 0], data0[:, 1], marker="o", label="target 0")
ax.scatter(data1[:, 0], data1[:, 1], marker="x", label="target 1")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```

```python
from sklearn.svm import SVC
```

```python
def plot_boundary_margin_sv(
    data0, target0, data1, target1, kernel, C, xmin=-1, xmax=1, ymin=-1, ymax=1
):
    svc = SVC(kernel=kernel, C=C)  # <-- 1
    svc.fit(np.vstack((data0, data1)), np.hstack((target0, target1)))  # <-- 2
    # 3
    fig, ax = plt.subplots()
    ax.scatter(data0[:, 0], data0[:, 1], marker="o", label="target 0")
    ax.scatter(data1[:, 0], data1[:, 1], marker="x", label="target 1")
    # 4
    xx, yy = np.meshgrid(np.linspace(xmin, xmax, 100), np.linspace(ymin, ymax, 100))
    xy = np.vstack([xx.ravel(), yy.ravel()]).T
    p = svc.decision_function(xy).reshape((100, 100))
    ax.contour(
        xx, yy, p, colors="k", levels=[-1, 0, 1], alpha=0.5, linestyles=[":", "-", ":"]
    )
    ax.scatter(
        svc.support_vectors_[:, 0],
        svc.support_vectors_[:, 1],
        s=250,
        facecolors="none",
        edgecolors="red",
    )  # <-- 5

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.legend(loc="best")
    plt.show()
```

```python
plot_boundary_margin_sv(data0, target0, data1, target1, kernel="linear", C=1e6)
```

```python
plot_boundary_margin_sv(data0, target0, data1, target1, kernel="linear", C=0.1)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

image = np.full((120, 180, 3), 0.9)
boxes = [
    {"label": "object A", "x": 20, "y": 25, "width": 60, "height": 45},
    {"label": "object B", "x": 105, "y": 40, "width": 45, "height": 55},
]

fig, ax = plt.subplots()
ax.imshow(image)
for box in boxes:
    patch = Rectangle(
        (box["x"], box["y"]),
        box["width"],
        box["height"],
        fill=False,
        edgecolor="red",
        linewidth=2,
    )
    ax.add_patch(patch)
    ax.text(box["x"], box["y"] - 3, box["label"], color="red")

ax.set_title("Bounding-box review")
ax.axis("off")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

annotations = pd.DataFrame({"box_area": [980, 1050, 1100, 1150, 1200, 1250, 9000]})
q1 = annotations["box_area"].quantile(0.25)
q3 = annotations["box_area"].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
annotations["review_candidate"] = ~annotations["box_area"].between(lower, upper)

fig, ax = plt.subplots()
ax.boxplot(annotations["box_area"], vert=False)
ax.set(xlabel="Bounding-box area (pixels²)", title="Candidate outlier review")
plt.show()

annotations
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

reviews = pd.DataFrame(
    {
        "label": ["A", "A", "A", "B", "B", "C", "C", "C"],
        "status": ["correct", "correct", "incorrect", "correct", "incorrect", "correct", "incorrect", "incorrect"],
    }
)
review_matrix = pd.crosstab(reviews["label"], reviews["status"])

sns.heatmap(review_matrix, annot=True, fmt="d", cmap="Blues")
plt.title("Review outcomes by label")
plt.show()
```

```python
import pandas as pd

annotations = pd.DataFrame(
    {
        "annotation_id": [1, 2, 2, 4],
        "label": ["cat", "dog", "dog", None],
        "x_min": [10, 20, -3, 15],
        "x_max": [50, 18, 40, 70],
    }
)
allowed_labels = {"cat", "dog"}

validation = pd.DataFrame(index=annotations.index)
validation["duplicate_id"] = annotations["annotation_id"].duplicated(keep=False)
validation["missing_label"] = annotations["label"].isna()
validation["unknown_label"] = ~annotations["label"].isin(allowed_labels) & annotations["label"].notna()
validation["invalid_box"] = (annotations["x_min"] < 0) | (annotations["x_max"] <= annotations["x_min"])

validation
```

```python
from sklearn.metrics import classification_report, cohen_kappa_score

reference = ["cat", "dog", "cat", "bird", "dog"]
annotator = ["cat", "dog", "dog", "bird", "dog"]

print(classification_report(reference, annotator, zero_division=0))
print("Cohen's kappa:", cohen_kappa_score(reference, annotator))
```

```python
import pandas as pd
from sklearn.metrics import classification_report, cohen_kappa_score, confusion_matrix

review = pd.DataFrame(
    {
        "reference": ["cat", "dog", "cat", "bird", "dog", "bird"],
        "annotation": ["cat", "dog", "dog", "bird", "dog", "cat"],
    }
)
labels = ["bird", "cat", "dog"]

print(classification_report(review["reference"], review["annotation"], labels=labels, zero_division=0))
print("Cohen's kappa:", cohen_kappa_score(review["reference"], review["annotation"]))

confusion = pd.DataFrame(
    confusion_matrix(review["reference"], review["annotation"], labels=labels),
    index=[f"reference_{label}" for label in labels],
    columns=[f"annotation_{label}" for label in labels],
)
confusion
```
