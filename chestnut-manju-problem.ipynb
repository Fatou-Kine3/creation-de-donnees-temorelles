```python
import math

# Assumptions
manju_volume = 100e-6  # 100 cm³ = 0.0001 m³
growth_time = 5        # minutes

# Solar system: approximate volume of the heliosphere
au = 149_597_870_700   # 1 AU in meters
radius = 123 * au

solar_system_volume = (4 / 3) * math.pi * radius**3

print("Manju volume:", manju_volume, "m³")
print("Solar system volume:", solar_system_volume, "m³")
```

```python
volume = manju_volume
minutes = 0

while volume < solar_system_volume:
    volume *= 2
    minutes += 5

print("Time required:", minutes, "minutes")
print("Time required:", minutes / 60, "hours")
print("Time required:", minutes / (60 * 24), "days")
```

```python
def calculate_growth_time(initial_volume, target_volume, doubling_time):
    volume = initial_volume
    time = 0

    while volume < target_volume:
        volume *= 2
        time += doubling_time

    return time
```

```python
time = calculate_growth_time(
    manju_volume,
    solar_system_volume,
    5
)

print("Time:", time, "minutes")
```

```python
tokyo_dome_volume = 1.24e6

time = calculate_growth_time(
    manju_volume,
    tokyo_dome_volume,
    5
)

print("Time to fill Tokyo Dome:", time, "minutes")
print("Time to fill Tokyo Dome:", time / 60, "hours")
```

```python
import matplotlib.pyplot as plt

volumes = [manju_volume]
times = [0]

volume = manju_volume
time = 0

while volume < solar_system_volume:
    volume *= 2
    time += 5

    volumes.append(volume)
    times.append(time)

plt.plot(times, volumes)
plt.xlabel("Time (minutes)")
plt.ylabel("Volume (m³)")
plt.title("Chestnut Manju Exponential Growth")
plt.yscale("log")
plt.show()
```

```python
import math

manju_volume = 100e-6
doubling_time = 5

print("Manju volume:", manju_volume, "m³")
```

```python
au = 149_597_870_700
radius = 123 * au

solar_system_volume = (4 / 3) * math.pi * radius**3

print("Solar system volume:", solar_system_volume, "m³")
```

```python
def calculate_growth_time(initial_volume, target_volume, doubling_time):
    n = math.ceil(math.log2(target_volume / initial_volume))
    return n * doubling_time
```

```python
time = calculate_growth_time(
    manju_volume,
    solar_system_volume,
    doubling_time
)

print("Time:", time, "minutes")
print("Time:", time / 60, "hours")
print("Time:", time / (60 * 24), "days")
```

```python
import matplotlib.pyplot as plt

volumes = [manju_volume]
times = [0]

volume = manju_volume
time = 0

while volume < solar_system_volume:
    volume *= 2
    time += doubling_time

    volumes.append(volume)
    times.append(time)

plt.plot(times, volumes)
plt.xlabel("Time (minutes)")
plt.ylabel("Volume (m³)")
plt.title("Chestnut Manju Exponential Growth")
plt.yscale("log")
plt.show()
```
